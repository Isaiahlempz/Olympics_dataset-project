```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Financial dataset
data = {
    'Month': ['Jan','Feb','Mar','Apr','May'],
    'Revenue': [120,135,150,160,155],
    'Expenses': [80,85,95,100,98],
    'ProfitMargin': [33,37,35,38,37],
    'CustomerGrowth': [5,6,8,9,7],
    'MarketingIndex': [1.2,1.3,1.5,1.6,1.4]
}
df = pd.DataFrame(data)

# 1. Line plot for Revenue & Expenses
plt.plot(df['Month'], df['Revenue'], label='Revenue', marker='o')
plt.plot(df['Month'], df['Expenses'], label='Expenses', marker='o')
plt.title("Revenue vs Expenses")
plt.xlabel("Month")
plt.ylabel("Value")
plt.legend()
plt.show()

# 2. Bar plot for Profit Margin
sns.barplot(x='Month', y='ProfitMargin', data=df, color='skyblue')
plt.title("Profit Margin by Month")
plt.show()

# 3. Scatter plot for Marketing vs Customer Growth
sns.scatterplot(x='MarketingIndex', y='CustomerGrowth', data=df, hue='Month', s=100)
plt.title("Marketing Index vs Customer Growth")
plt.show()

# 4. Multivariate pairplot (relationships among all variables)
sns.pairplot(df[['Revenue','Expenses','ProfitMargin','CustomerGrowth','MarketingIndex']])
plt.show()
```

```python
import pandas as pd

data = {
    'Month':['Jan','Feb','Mar','Apr','May','Jun'],
    'PassengerCount':[112,118,132,129,121,135]
}
df = pd.DataFrame(data)

# Rolling average (window=3)
df['RollingAvg'] = df['PassengerCount'].rolling(window=3).mean()

# Percentage change
df['PctChange'] = df['PassengerCount'].pct_change() * 100

print(df)
```

```python
# Variance and standard deviation of passenger counts
var = df['PassengerCount'].var()
std = df['PassengerCount'].std()

# Correlation between PassengerCount and AvgTemp
data2 = {
    'PassengerCount':[112,118,132,129,121,135],
    'AvgTemp':[23,24,26,27,28,30]
}
df2 = pd.DataFrame(data2)

corr = df2['PassengerCount'].corr(df2['AvgTemp'])

print("Variance:", var)
print("Standard Deviation:", std)
print("Correlation (Passenger vs Temp):", corr)
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# Dataset
data = {
    'ID': ['M001','M002','M003','M004','M005','M006'],
    'Feature1':[5.1,4.9,6.2,5.9,5.6,5.7],
    'Feature2':[3.5,3.0,3.4,3.0,2.9,2.8],
    'Feature3':[1.4,1.4,5.4,5.1,3.6,4.1],
    'Feature4':[0.2,0.2,2.3,1.8,1.3,1.3],
    'Label':['Setosa','Setosa','Virginica','Virginica','Versicolor','Versicolor']
}
df = pd.DataFrame(data)

# Features and target
X = df[['Feature1','Feature2','Feature3','Feature4']]
y = df['Label']

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train Decision Tree
clf = DecisionTreeClassifier()
clf.fit(X_train, y_train)

# Predict
y_pred = clf.predict(X_test)

print("Predictions:", y_pred)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
import pandas as pd
import numpy as np
from scipy import stats

# Patient Vitals Dataset
data1 = {
    'ID':['P001','P002','P003','P004','P005','P006'],
    'Pulse':[72,85,60,78,65,90],
    'Systolic':[120,135,110,125,115,140],
    'Diastolic':[80,85,70,78,75,90],
    'Cholesterol':[190,220,180,200,210,230],
    'AgeGroup':['Adult','Adult','Senior','Adult','Senior','Adult']
}
df1 = pd.DataFrame(data1)

# (a) Pulse statistics
print("Pulse Mean:", df1['Pulse'].mean())
print("Pulse Median:", df1['Pulse'].median())
print("Pulse Variance:", df1['Pulse'].var(ddof=0))  # population variance
print("Pulse Std Dev:", df1['Pulse'].std(ddof=0))

# (b) Covariance matrix + eigen decomposition
cov_matrix = np.cov(df1[['Pulse','Systolic','Diastolic','Cholesterol']].T)
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
print("Covariance Matrix:\n", cov_matrix)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:\n", eigenvectors)

# (c) Cholesterol variability
print("Cholesterol Variance:", df1['Cholesterol'].var(ddof=0))
print("Cholesterol IQR:", df1['Cholesterol'].quantile(0.75) - df1['Cholesterol'].quantile(0.25))

# (d) Pulse anomalies using Z-score
df1['Pulse_Z'] = stats.zscore(df1['Pulse'])
print(df1[['ID','Pulse','Pulse_Z']])
```

```python
# E-Commerce Dataset
data2 = {
    'ID':['E001','E002','E003','E004','E005','E006'],
    'Month':['Jan','Feb','Mar','Apr','May','Jun'],
    'UnitsSold':[200,230,210,250,220,260],
    'Promotion':['No','Yes','No','Yes','No','Yes'],
    'Revenue':[4000,4600,4200,5000,4400,5200],
    'Customers':[120,140,130,150,135,160]
}
df2 = pd.DataFrame(data2)

# (a) Revenue growth rates
df2['RevenueGrowth'] = df2['Revenue'].pct_change() * 100
print(df2[['Month','Revenue','RevenueGrowth']])

# (b) Dispersion measures
print("Customers Variance:", df2['Customers'].var(ddof=0))
print("Customers Std Dev:", df2['Customers'].std(ddof=0))
print("Revenue Variance:", df2['Revenue'].var(ddof=0))
print("Revenue Std Dev:", df2['Revenue'].std(ddof=0))

# (c) Pearson correlation UnitsSold vs Revenue
corr = df2['UnitsSold'].corr(df2['Revenue'])
print("Correlation UnitsSold vs Revenue:", corr)
```

```python
# Energy Consumption Dataset
data3 = {
    'ID':['EN001','EN002','EN003','EN004','EN005','EN006'],
    'Temp':[24,28,22,27,23,26],
    'Humidity':[60,55,70,58,72,62],
    'EnergyUsage':[350,400,300,420,320,380],
    'WindSpeed':[10,12,8,11,7,9],
    'Location':['Residential','Residential','Industrial','Residential','Industrial','Residential']
}
df3 = pd.DataFrame(data3)

# (a) Feature engineering
df3['Temp_Humidity'] = df3['Temp'] * df3['Humidity']   # interaction
df3['Usage_per_Temp'] = df3['EnergyUsage'] / df3['Temp']  # ratio
df3['Temp_squared'] = df3['Temp']**2                   # polynomial
print(df3[['Temp','Humidity','EnergyUsage','Temp_Humidity','Usage_per_Temp','Temp_squared']])

# (b) Grouped statistics by Location
group_stats = df3.groupby('Location')['EnergyUsage'].mean()
print("Mean EnergyUsage by Location:\n", group_stats)

# (c) Regression model
import statsmodels.api as sm

X = df3[['Temp','Humidity','WindSpeed']]
y = df3['EnergyUsage']
X = sm.add_constant(X)  # add intercept
model = sm.OLS(y, X).fit()
print(model.summary())
```
